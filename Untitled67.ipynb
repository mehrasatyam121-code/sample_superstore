```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/Sample - Superstore.csv", encoding='latin1')
```

```python
df.head(5)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'])
```

```python
df
```

```python

total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()

print(f"Total Sales: {total_sales:.2f}")
print(f"Total Profit: {total_profit:.2f}")
```

```python
top_10_customers_by_sales = df.groupby('Customer Name')['Sales'].sum().nlargest(10)
display(top_10_customers_by_sales)
```

```python
df['Order_Month_Year'] = df['Order Date'].dt.to_period('M')
monthly_sales = df.groupby('Order_Month_Year')['Sales'].sum().reset_index()
monthly_sales['Order_Month_Year'] = monthly_sales['Order_Month_Year'].dt.to_timestamp()
```

```python
plt.figure(figsize=(14, 7))
sns.lineplot(x='Order_Month_Year', y='Sales', data=monthly_sales)
plt.title('Monthly Sales Trend')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='Category', y='Profit', data=category_profit, palette='viridis')
plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
correlation_sales_profit = df['Sales'].corr(df['Profit'])
print(f"Correlation between Sales and Profit: {correlation_sales_profit:.2f}")
```

### Sales by Region

```python
sales_by_region = df.groupby('Region')['Sales'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='Region', y='Sales', data=sales_by_region, palette='coolwarm')
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.tight_layout()
plt.show()
```

### Profit by Segment

```python
profit_by_segment = df.groupby('Segment')['Profit'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='Segment', y='Profit', data=profit_by_segment, palette='plasma')
plt.title('Total Profit by Segment')
plt.xlabel('Segment')
plt.ylabel('Total Profit')
plt.tight_layout()
plt.show()
```

### Sales vs. Profit Scatter Plot

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Sales', y='Profit', data=df, hue='Category', palette='deep', alpha=0.6)
plt.title('Sales vs. Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
